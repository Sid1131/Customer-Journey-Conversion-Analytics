## EDA

```python
continuous_cols = [
    'administrative_duration',
    'informational_duration',
    'productrelated_duration',
    'bouncerates',
    'exitrates',
    'pagevalues',
    'specialday'
]
```

```python
sns.set_style('whitegrid')

for col in continuous_cols:
    plt.figure(figsize=(8,5))
    sns.histplot(data=df, x=col, bins=30, kde=True)

    plt.title(f'Distribution of {col}')
    plt.show()
```

1. Administrative_Duration
Observation:
    1. The distribution is highly right-skewed.
    2. Most visitors spend very little time on administrative pages.
    3. A small number of users spend significantly longer durations, creating a long right tail.
    4. The presence of extreme values suggests potential outliers, which are acceptable for tree-based models like Random Forest and XGBoost.

2. Informational_Duration
Observation:
   1. The distribution is strongly positively skewed.
   2. Most sessions contain little or no time spent on informational pages.
   3. A few visitors spend considerably more time, indicating occasional deep information-seeking behavior.
   4. Several high-duration observations appear as potential outliers.

3. ProductRelated_Duration
Observation:
    1. The feature exhibits an extreme positive skew with a long right tail.
    2. Most users spend relatively little time on product-related pages.
    3. A small group of users spends substantially longer browsing products.
    4. These longer browsing sessions may indicate higher purchase intent and could be important predictors of revenue.

4. BounceRates
Observation:
    1. Bounce rates are concentrated near zero, indicating that many users continue browsing beyond the landing page.
    2. The distribution is highly right-skewed.
    3. A noticeable spike exists around 0.20, representing sessions with relatively high bounce rates.
    4. Higher bounce rates may be negatively associated with purchase completion.

5. ExitRates
Observation:
    1. Exit rates are moderately right-skewed.
    2. Most observations lie between 0.01 and 0.05.
    3. A secondary concentration near 0.20 suggests some sessions end with relatively high exit probabilities.
    4. Exit rate is expected to be an important predictor of purchasing behavior.

6. PageValues
Observation:
    1. The majority of observations have PageValues close to zero.
    2. The feature is highly right-skewed with a few very large values.
    3. High PageValues typically indicate pages that contribute significantly to conversions.
    4. This feature is expected to have strong predictive power for the target variable.

7. SpecialDay
Observation:
    1. Most sessions occurred when SpecialDay = 0, indicating they were not close to a special shopping event.
    2. Only a small proportion of observations correspond to values between 0.2 and 1.0.
    3. The feature is highly imbalanced and may have limited standalone predictive power, though it could still contribute when combined with other         features.

Overall Observation:

1. Most continuous features exhibit positive (right) skewness, with a large concentration of observations near lower values and a small number of high-value observations.
2. Several variables contain potential outliers, particularly the duration-based features and PageValues. Since tree-based algorithms such as Random Forest and XGBoost are robust to outliers, these values will be retained.
3. Features such as ProductRelated_Duration, PageValues, BounceRates, and ExitRates appear particularly informative and are expected to play an  important role in predicting customer purchase intention.

```python
categorical_columns = ['weekend', 'revenue', 'month', 'visitortype']
for cols in categorical_columns:
    plt.figure(figsize=(8,6))
    sns.countplot(
        data=df,
        x=cols
    )
    plt.title(f'Distibution of {cols}')
    plt.xticks(rotation=45)
    plt.show()
```

Observations:
1. Weekend:
    The majority of website sessions occur on weekdays (Weekend = False), indicating that users are more likely to browse the website during regular      working days than on weekends.
2. Revenue:
    Most website sessions do not result in a purchase (Revenue = False), indicating that only a small proportion of visitors complete a transaction.     This also suggests that the target variable is imbalanced.
3. Month:
    Website traffic is highest during May and November, indicating seasonal variations in customer activity. These months may correspond to      f       promotional events or increased shopping demand.
4. VisitorType:
  The majority of website visitors are Returning Visitors, suggesting that repeat customers constitute a significant portion of the website traffic    and may have higher purchase intent than new visitors.

# Bivariate Analysis : Continous Features vs Revenue

```python
for col in continuous_cols:
    plt.figure(figsize=(8,5))
    sns.boxplot(
        data=df,
        x='revenue',
        y=col
    )
    plt.title(f'{col} vs Revenue')
    plt.show()
```

Observations
1. Administrative_Duration:
Purchasing customers generally spend more time on administrative pages than non-purchasing customers.

2. Informational_Duration:
Customers who make a purchase tend to spend slightly more time on informational pages.

3. ProductRelated_Duration:
Purchasing customers spend significantly more time on product-related pages, indicating higher buying intent.

4. BounceRates:
Customers who generate revenue have lower bounce rates, suggesting better engagement.

5. ExitRates:
Lower exit rates are associated with customers who complete a purchase.

6. PageValues:
Purchasing customers have considerably higher PageValues, making it one of the strongest indicators of revenue.

7. SpecialDay:
Most sessions occur when SpecialDay = 0, and no clear relationship with revenue is observed.

# Bivariate analysis : Categorical Columns vs Revenue

```python
categorical_cols = ['weekend', 'month', 'visitortype']
for cols in categorical_cols:
    plt.figure(figsize=(8,6))
    sns.countplot(
        data=df,
        x=cols,
        hue='revenue'
    )
    plt.title(f'{cols} vs Revenue')
    plt.show()
```

Observations:

2. Month : Highest revenue generations comes from the month of 'NOVERMBER' , 'DECEMBER' and 'MAY'.

Observations:

2. Browser:
Browser 2 accounts for the highest number of website visits and purchases, whereas other browsers have comparatively lower user activity.
3. Region:
Region 1 has the highest website traffic and purchases, although purchases are observed across all regions.
4. TrafficType:
Traffic Types 2, 1, and 3 generate the highest number of visitors and purchases, while the remaining traffic sources contribute relatively less traffic.
5. Overall Observation
User activity and purchases are concentrated in a few operating systems, browsers, regions, and traffic sources, indicating that these features may provide useful information for predicting customer purchase behavior.

```python
plt.figure(figsize=(12,8))
corr = df.corr(numeric_only=True)
sns.heatmap(
    corr,
    annot=True,
    fmt='.2f',
    linewidth=0.5,
    square=True
)
plt.title("Correlation Heatmap")
plt.show()
```

Observations:
1. Most features exhibit weak to moderate correlations, indicating that multicollinearity is generally not a major concern.
2. ProductRelated and ProductRelated_Duration have a strong positive correlation (0.86), suggesting that users who visit more product pages also tend     to spend more time on them.
3. BounceRates and ExitRates show a very strong positive correlation (0.90), indicating that both features capture similar aspects of user engagement.
4. PageValues has the highest positive correlation with Revenue (0.49), making it one of the strongest predictors of customer purchases.
5. BounceRates (-0.15) and ExitRates (-0.20) are negatively correlated with Revenue, suggesting that higher bounce and exit rates reduce the   r  r rt    likelihood of a purchase.
6. ProductRelated (0.16), ProductRelated_Duration (0.15), and Administrative (0.14) have a weak positive correlation with Revenue, indicating that  d     greater engagement with the website slightly increases purchase probability.
7. SpecialDay, OperatingSystems, Browser, Region, TrafficType, and Weekend show very weak correlations with Revenue, implying they have limited   f f     individual predictive power.
